```python
print_config_summary()
```

```python
env = DynamicPricingEnv()

print("Training agents for benchmarks...\n")

ppo = PPOAgent(env, {**BEST_PPO_CONFIG,
                     'n_episodes': 200})
ppo.train(n_episodes=200, verbose=False)
print("✅ PPO ready")

dqn = DQNAgent(env, {**BEST_DQN_CONFIG,
                     'n_episodes': 200})
dqn.train(n_episodes=200, verbose=False)
print("✅ DQN ready")

ql = QLearningAgent(env, QL_CONFIG)
ql.train(n_episodes=500, verbose=False)
print("✅ Q-Learning ready")
```

```python
agents_to_bench = {
    'PPO'         : ppo,
    'DQN'         : dqn,
    'Q-Learning'  : ql,
    'Time Based'  : TimedPricingAgent(env),
    'Demand Based': DemandBasedAgent(env),
}

print("Running benchmarks...")
bench_results = benchmark_agents(
    agents_to_bench, env,
    n_episodes=50
)
```
